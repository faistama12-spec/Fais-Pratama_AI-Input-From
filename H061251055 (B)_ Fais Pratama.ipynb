# 5 Prompt Geofisika dengan Struktur R–T–C–I–K–O–Q–V
## Struktur Prompt

- **T — Task**: Tugas atau apa yang harus dikerjakan

- **Q — Quality Criteria**: Kriteria kualitas hasil

# Prompt 1 — Pemodelan Inversi Geolistrik

```markdown
# R — Role
Anda berperan sebagai ahli geofisika dekat permukaan (near-surface geophysics) dan instruktur pengolahan data geolistrik.

# T — Task
Lakukan interpretasi kurva resistivitas semu hasil pengukuran sounding geolistrik 1D untuk memperkirakan ketebalan dan nilai resistivitas lapis batuan di bawah permukaan.

# C — Context
Metode geolistrik konfigurasi Schlumberger sering digunakan dalam pencarian air tanah (akuifer). Pemodelan inversi diperlukan untuk mencocokkan data lapangan dengan model teoretis.

# I — Input
Data lapangan berupa spasi elektroda ($AB/2$) dan nilai resistivitas semu ($\rho_a$) yang tercatat pada lembar data praktikum.

# K — Constraints
- Gunakan asumsi perlapisan horizontal homogen.
- Batasi iterasi inversi sesuai dengan kewajaran data lapangan.
- Jangan mengabaikan titik data penunjang tanpa alasan valid.
- Jelaskan kaitan antara nilai resistivitas dengan jenis litologi umum.
- Tunjukkan nilai RMS error dari pemodelan.

# O — Output
Berikan:
1. Grafik pencocokan kurva (curve matching).
2. Tabel model lapisan bumi (ketebalan dan resistivitas)
3. Nilai RMS error inversi.
4. Interpretasi litologi lapisan bawah permukaan.
5. Kesimpulan keberadaan akuifer.

# Q — Quality Criteria
Hasil pemodelan harus akurat, nilai RMS error sekecil mungkin, dan interpretasi litologi harus sesuai dengan kondisi geologi lokal.

# V — Verification
Periksa kembali kesesuaian kurva lapangan dan kurva teoretis serta pastikan satuan resistivitas ($\Omega\text{m}$) sudah benar.
```

# Prompt 2 — Analisis Tomografi Seismik Refraksi

```markdown
# R — Role
Anda berperan sebagai ahli seismologi teknik dan pengembang model kecepatan gelombang.

# T — Task
Analisis waktu tiba (first break) gelombang refraksi untuk merekonstruksi penampang 2D kecepatan Vp bawah permukaan pada perencanaan pondasi teknik sipil.

# C — Context
Tomografi seismik refraksi digunakan untuk mengetahui tingkat pelapukan batuan dan kedalaman batuan dasar (bedrock) pada lokasi pembangunan infrastruktur.

# I — Input
Data waktu tiba first break dari berbagai offset geofon dan geometri lintasan pengukuran penembakan (shot gather).

# K — Constraints
- Identifikasi waktu first break secara konsisten.
- Perhatikan zona kecepatan rendah jika ada.
- Jangan melakukan smoothing berlebihan yang menghilangkan anomali lokal.
- Gunakan satuan jarak meter dan kecepatan meter per detik.
- Jelaskan batasan resolusi model tomografi.

# O — Output
Hasil harus mencakup:
1. Kurva travel-time.
2. Penampang model kecepatan 2D.
3. Estimasi kedalaman bedrock.
4. Penilaian tingkat kekerasan batuan.
5. Interpretasi geoteknik.

# Q — Quality Criteria
Penampang kecepatan harus logis secara fisik, waktu residual kecil, dan interpretasi aman untuk rekomendasi rekayasa teknik.

# V — Verification
Validasi model kecepatan menggunakan metode ray tracing ulang dan pastikan konsisten dengan data travel-time observasi.
```

# Prompt 3 — Analisis Penentuan Hiposenter Gempa

```markdown
# R — Role
Anda berperan sebagai peneliti pusat gempa bumi dan analis mitigasi bencana.

# T — Task
Tentukan koordinat episenter dan kedalaman hiposenter suatu kejadian gempa bumi menggunakan metode lingkaran konvergen dari beberapa stasiun pemantau.

# C — Context
Penentuan parameter sumber gempa secara cepat dan tepat sangat krusial dalam analisis tingkat bahaya guncangan serta peringatan dini tsunami.

# I — Input
Data waktu tiba gelombang P dan S dari minimal 3 stasiun pengamatan gempa beserta koordinat lintang dan bujur stasiun tersebut.

# K — Constraints
- Gunakan kecepatan rambat rata-rata gelombang kerak bumi.
- Hitung jarak episenter dengan rumus selisih waktu tiba.
- Gambarkan titik potong lingkaran dari stasiun secara proporsional.
- Abaikan stasiun yang mengalami gangguan sinyal berat (noise).
- Sajikan perhitungan secara sistematis.

# O — Output
Berikan:
1. Tabel selisih waktu P-S dan jarak stasiun.
2. Estimasi koordinat episenter (Lintang, Bujur).
3. Estimasi kedalaman hiposenter.
4. Analisis tingkat ketidakpastian posisi.
5. Peta skematik posisi episenter.

# Q — Quality Criteria
Perhitungan matematis harus presisi, titik konvergensi stasiun jelas, dan hasil analisis memenuhi standar pelaporan kegempaan.

# V — Verification
Periksa kembali lingkaran jarak dari setiap stasiun untuk memastikan tidak ada kesalahan konversi skala jarak.
```

# Prompt 4 — Koreksi dan Analisis Data Anomali Magnetik

```markdown
# R — Role
Anda berperan sebagai ahli survei magnetik mineral dan eksplorasi bijih besi.

# T — Task
Lakukan reduksi data medan magnet total lapangan meliputi koreksi harian (diurnal correction) dan koreksi IGRF untuk mendapatkan peta anomali magnetik sisa.

# C — Context
Variasi harian medan magnet bumi dan pengaruh medan utama bumi (IGRF) harus dihilangkan agar anomali yang murni berasal dari kemagnetan batuan lokal dapat dianalisis.

# I — Input
Data pembacaan base station, data pengukuran roving magnetometer, serta nilai parameter IGRF berdasarkan posisi dan tanggal pengukuran.

# K — Constraints
- Sinkronkan waktu pencatatan base station dan stasiun ukur.
- Terapkan nilai koreksi IGRF sesuai tahun survei.
- Jangan mengubah data mentah selain proses koreksi standar.
- Jelaskan tahapan matematis koreksi diurnal.
- Gunakan satuan nanoTesla (nT).

# O — Output
Tampilkan:
1. Grafik koreksi variasi harian.
2. Nilai koreksi IGRF.
3. Tabel hasil perhitungan anomali magnetik total dan sisa.
4. Sebaran nilai anomali maksimum dan minimum.
5. Analisis awal struktur sumber anomali.

# Q — Quality Criteria
Proses koreksi harus berurutan, bebas dari kesalahan hitung waktu, dan menghasilkan nilai anomali sisa yang valid secara geofisika.

# V — Verification
Periksa konsistensi nilai anomali sisa terhadap tren regional serta pastikan tidak ada data outlier yang terlewat.
```

# Prompt 5 — Analisis Zona Rawan Longsor Menggunakan SIG dan Parameter Geofisika

```markdown
# R — Role
Anda berperan sebagai analis kebencanaan geospasial dan ahli mitigasi tanah longsor.

# T — Task
Lakukan overlay beberapa parameter peta tematik (kemiringan lereng, curah hujan, jenis tanah, dan ketebalan soil pelapukan geofisika) untuk menghasilkan peta zona kerawanan longsor.

# C — Context
Kombinasi data parameter permukaan dari SIG dan ketebalan lapisan lapuk bawah permukaan dari data geofisika sangat membantu dalam menentukan kestabilan lereng secara komprehensif.

# I — Input
Data raster DEM, peta curah hujan wilayah, peta jenis tanah, serta data hasil estimasi kedalaman zona pelapukan batuan.

# K — Constraints
- Lakukan reproyeksi seluruh data raster ke sistem koordinat yang seragam.
- Berikan bobot (weighting) yang objektif pada tiap parameter overlay.
- Gunakan tool Weighted Overlay di perangkat lunak SIG.
- Validasi kelas kerawanan dengan titik kejadian longsor historis.
- Jangan menyederhanakan klasifikasi tanpa dasar ilmiah.

# O — Output
Berikan:
1. Matriks pembobotan parameter kerawanan.
2. Tahapan analisis overlay di perangkat SIG.
3. Peta zonasi kerawanan longsor (klasifikasi rendah, sedang, tinggi).
4. Persentase luas wilayah untuk tiap zona kerawanan.
5. Rekomendasi mitigasi bencana.

# Q — Quality Criteria
Analisis spasial harus konsistent, tahapan pengerjaan sistematis, dan pembobotan parameter sesuai dengan literatur kebencanaan geologi.

# V — Verification
Periksa ulang matriks bobot total agar bernilai 100% dan pastikan legenda peta akhir terbaca dengan jelas.
```
